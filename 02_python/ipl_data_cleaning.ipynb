## Data Cleaning

```python
# MAKE COPIES

matches_clean=matches.copy()
deliveries_clean=deliveries.copy()
print("Copies created successfully")
```

```python
# STANDARDIZWD COLUMN NAMES
matches_clean.columns=(matches_clean.columns.str.strip().str.lower().str.replace(" ","_"))
deliveries_clean.columns=(deliveries_clean.columns.str.strip().str.lower().str.replace(" ","_"))

print(matches_clean.columns.tolist())
print(deliveries_clean.columns.tolist())
```

```python
# REMOVE UNNECESSARY SPACES
```

```python
for df in [matches_clean,deliveries_clean]:
    text_columns=df.select_dtypes(include="object").columns
    for col in text_columns:
        df[col]=df[col].str.strip()

print("Text Field Standaerdized")
```

```python
# STANDARDIZE TEAM NAMES

team_mapping = {
    "Delhi Daredevils": "Delhi Capitals",
    "Kings XI Punjab": "Punjab Kings",
    "Rising Pune Supergiants": "Rising Pune Supergiant",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru"
}
```

```python
matches_clean["team1"] = matches_clean["team1"].replace(team_mapping)
matches_clean["team2"] = matches_clean["team2"].replace(team_mapping)
matches_clean["toss_winner"] = matches_clean["toss_winner"].replace(team_mapping)
matches_clean["winner"] = matches_clean["winner"].replace(team_mapping)
deliveries_clean["batting_team"] = deliveries_clean["batting_team"].replace(team_mapping)
deliveries_clean["bowling_team"] = deliveries_clean["bowling_team"].replace(team_mapping)
```

```python
# STANDERDIZED VENUE NAMES
sorted(matches["venue"].dropna().unique())
```

```python
venue_mapping={"Arun Jaitley Stadium":"Arun Jaitley Stadium, Delhi",
               "Brabourne Stadium":"Brabourne Stadium,Mumbai",
               "Dr DY Patil Sports Academy":"Dr DY Patil Sports Academy, Mumbai",
               "Dr. Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium":"Dr. Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium, Visakhapatnam",
               "Eden Gardens":"Eden Gardens, Kolkata",
               "Himachal Pradesh Cricket Association Stadium":"Himachal Pradesh Cricket Association Stadium, Dharamsala",
               "M Chinnaswamy Stadium":"M Chinnaswamy Stadium, Bengaluru",
               "M.Chinnaswamy Stadium":"M Chinnaswamy Stadium, Bengaluru",
               "MA Chidambaram Stadium, Chepauk":"MA Chidambaram Stadium, Chepauk, Chennai",
               "MA Chidambaram Stadium":"MA Chidambaram Stadium, Chepauk, Chennai",
               "Maharashtra Cricket Association Stadium":"Maharashtra Cricket Association Stadium, Pune",
               "Punjab Cricket Association Stadium, Mohali":"Punjab Cricket Association IS Bindra Stadium, Mohali, Chandigarh",
               "Punjab Cricket Association IS Bindra Stadium, Mohali":"Punjab Cricket Association IS Bindra Stadium, Mohali, Chandigarh",
               "Punjab Cricket Association IS Bindra Stadium":"Punjab Cricket Association IS Bindra Stadium, Mohali, Chandigarh",
               "Rajiv Gandhi International Stadium, Uppal":"Rajiv Gandhi International Stadium, Uppal, Hyderabad",
               "Rajiv Gandhi International Stadium":"Rajiv Gandhi International Stadium, Uppal, Hyderabad",
               "Wankhede Stadium":"Wankhede Stadium, Mumbai",
               "Sawai Mansingh Stadium":"Sawai Mansingh Stadium, Jaipur",
               "Sheikh Zayed Stadium":"Zayed Cricket Stadium, Abu Dhabi",
               }
```

```python
matches_clean["venue"] = matches_clean["venue"].replace(venue_mapping)
```

```python
sorted(matches_clean["venue"].dropna().unique())
```

```python
sorted(matches_clean["city"].dropna().unique())
```

```python
# CONVERY DATYPE INTO DATE

matches_clean["date"] = pd.to_datetime(
    matches_clean["date"],
    errors="coerce"
)

print("Date type:", matches_clean["date"].dtype)
```

```python
# CHECK DUPLICATES

print("Duplicate match IDs:",
      matches_clean["id"].duplicated().sum())

print("Duplicate match rows:",
      matches_clean.duplicated().sum())

print("Duplicate delivery rows:",
      deliveries_clean.duplicated().sum())
```

```python
matches_clean.isnull().sum()
```

```python
# HANDLI MISSING CITY 

matches_clean[matches_clean["city"].isna()][
    ["id", "venue", "city"]
]
```

```python
venue_city_mapping = {
    "Sharjah Cricket Stadium": "Sharjah",
    "Dubai International Cricket Stadium": "Dubai"
}

matches_clean["city"] = (
    matches_clean["city"]
    .fillna(matches_clean["venue"].map(venue_city_mapping))
)

matches_clean["method_status"] = (
    matches_clean["method"].fillna("Normal")
)
```

```python
matches_clean["city"].isna().sum()
```

###  Winner validation

```python
invalid_winners = matches_clean[
    matches_clean["winner"].notna() &
    ~(
        (matches_clean["winner"] == matches_clean["team1"]) |
        (matches_clean["winner"] == matches_clean["team2"])
    )
]

print("Invalid winner records:", len(invalid_winners))
```

### Toss winner validation

```python
invalid_toss_winners = matches_clean[
    matches_clean["toss_winner"].notna() &
    ~(
        (matches_clean["toss_winner"] == matches_clean["team1"]) |
        (matches_clean["toss_winner"] == matches_clean["team2"])
    )
]

print("Invalid toss winner records:", len(invalid_toss_winners))
```

### Referencial intigrity

```python
match_ids = set(matches_clean["id"])

invalid_delivery_matches = deliveries_clean[
    ~deliveries_clean["match_id"].isin(match_ids)
]

print(
    "Deliveries with invalid match IDs:",
    len(invalid_delivery_matches)
)
```

### Missing value report

```python
missing_report = pd.DataFrame({
    "Column": matches_clean.columns,
    "Missing_Count": matches_clean.isnull().sum().values,
    "Missing_Percentage": (
        matches_clean.isnull().mean().values * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing_Count",
    ascending=False
)

missing_report
```

### Delivery missing value report

```python
delivery_missing_report = pd.DataFrame({
    "Column": deliveries_clean.columns,
    "Missing_Count": deliveries_clean.isnull().sum().values,
    "Missing_Percentage": (
        deliveries_clean.isnull().mean().values * 100
    ).round(2)
})

delivery_missing_report = delivery_missing_report.sort_values(
    by="Missing_Count",
    ascending=False
)

delivery_missing_report
```

## Final quality summary

```python
quality_summary = pd.DataFrame({
    "Dataset": ["Matches", "Deliveries"],
    "Rows": [
        len(matches_clean),
        len(deliveries_clean)
    ],
    "Columns": [
        len(matches_clean.columns),
        len(deliveries_clean.columns)
    ],
    "Duplicate_Rows": [
        matches_clean.duplicated().sum(),
        deliveries_clean.duplicated().sum()
    ],
    "Missing_Cells": [
        matches_clean.isnull().sum().sum(),
        deliveries_clean.isnull().sum().sum()
    ]
})

quality_summary
```

```python
# FINAL QUALITY CHECK

final_quality = pd.DataFrame({
    "Check": [
        "Total Matches",
        "Total Deliveries",
        "Duplicate Match IDs",
        "Duplicate Match Rows",
        "Duplicate Delivery Rows",
        "Invalid Winners",
        "Invalid Toss Winners",
        "Invalid Delivery Match IDs",
        "Missing Dates",
        "Missing Cities"
    ],
    "Result": [
        len(matches_clean),
        len(deliveries_clean),
        matches_clean["id"].duplicated().sum(),
        matches_clean.duplicated().sum(),
        deliveries_clean.duplicated().sum(),
        len(invalid_winners),
        len(invalid_toss_winners),
        len(invalid_delivery_matches),
        matches_clean["date"].isna().sum(),
        matches_clean["city"].isna().sum()
    ]
})

final_quality
```
